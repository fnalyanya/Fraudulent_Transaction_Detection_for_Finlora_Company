VARIABLES WITH MISSING VALUES THAT NEED IMPUTATION

amount_usd: has 305 missing values. This will be filled by multiplying amount_src by the exchange rate of the currency.

fee: has 295 missing values. These will be filled using the median fee of each channel (if channel exists). Any remaining missing fee values will be filled using the overall median.

ip_country: has 301 missing values. These will be filled using the corresponding home_country.

kyc_tier: has 300 missing values. These will be filled using the most frequent kyc_tier (mode).

device_trust_score: has 295 missing values. These will be filled using the median device_trust_score grouped by kyc_tier and new_device. Remaining missing values will be filled using the overall median.

timestamp: has 61 missing values. These rows will be dropped because the missing proportion is very small.

amount_src: has 4 missing values. These rows will be dropped because the missing proportion is very small.

ip_address: has 305 missing values. These cannot be imputed because each IP address is unique. Rows with missing ip_address will be dropped.

```python
#calculating exchange rates per currency
exchange_rates = customer_transaction_data[customer_transaction_data['amount_usd'].notna()].groupby('source_currency').apply(
    lambda x: (x['amount_usd'] / x['amount_src']).mean()
).to_dict()
exchange_rates
```

```python
#filling the missing values of the amount_usd
customer_transaction_data['amount_usd'] = customer_transaction_data.apply(
    lambda row: row['amount_usd'] if pd.notna(row['amount_usd']) else row['amount_src'] * exchange_rates.get(row['source_currency'], 1),
    axis=1
)
```

```python
missing_df = pd.DataFrame({
    'Column Name': customer_transaction_data.columns,
    'Missing Count': customer_transaction_data.isna().sum(),
    'Missing Pct': customer_transaction_data.isna().mean() * 100
})

missing_df
```

```python
# filling the missing values of the fee variable
if 'fee' in customer_transaction_data.columns:
    if 'channel' in customer_transaction_data.columns:
        customer_transaction_data['fee'] = customer_transaction_data.groupby('channel')['fee'].transform(lambda y: y.fillna(y.median()))
    customer_transaction_data['fee'] = customer_transaction_data['fee'].fillna(customer_transaction_data['fee'].median())
```

```python
# filling the missing values of ip_country variable
if {'ip_country', 'home_country'}.issubset(customer_transaction_data.columns):
    customer_transaction_data['ip_country'] = customer_transaction_data['ip_country'].fillna(customer_transaction_data['home_country'])
```

```python
# filling the missing values of ip_country variable
if {'ip_country', 'home_country'}.issubset(customer_transaction_data.columns):
    customer_transaction_data['ip_country'] = customer_transaction_data['ip_country'].fillna(customer_transaction_data['home_country'])
```

```python
# filling the missing values of device_trust_score variable
if 'device_trust_score' in customer_transaction_data.columns:
    if {'new_device', 'kyc_tier'}.issubset(customer_transaction_data.columns):
        customer_transaction_data['device_trust_score'] = customer_transaction_data.groupby(['new_device', 'kyc_tier'])['device_trust_score'].transform(lambda x: x.fillna(x.mean()))
    customer_transaction_data['device_trust_score'] = customer_transaction_data['device_trust_score'].fillna(customer_transaction_data['device_trust_score'].mean())
```

```python
# dropping the timestamp, amount_src and ip_address missing values
customer_transaction_data.dropna(subset=['timestamp', 'amount_src', 'ip_address'], inplace=True)
```

```python
import pandas as pd

cleaned_customer_transaction_data = pd.read_csv('../Finlora_Dataset/FinLora_Customer_Transaction_Dataset.csv')

cleaned_customer_transaction_data.head()
cleaned_customer_transaction_data.info()

```

```python
# --- Normalize home_country ---

# Clean formatting
cleaned_customer_transaction_data['home_country'] = (
    cleaned_customer_transaction_data['home_country']
    .astype(str)
    .str.strip()
    .str.upper()
)

# Map ISO codes and variations to full names
home_country_map = {
    'UK': 'UNITED KINGDOM',
    'U.K.': 'UNITED KINGDOM',
    'GB': 'UNITED KINGDOM',
    'KE': 'KENYA',
    'US': 'UNITED STATES',
    'USA': 'UNITED STATES'
}

cleaned_customer_transaction_data['home_country'] = (
    cleaned_customer_transaction_data['home_country']
    .replace(home_country_map)
)
```

```python
# --- Normalize ip_country ---

cleaned_customer_transaction_data['ip_country'] = (
    cleaned_customer_transaction_data['ip_country']
    .astype(str)
    .str.strip()
    .str.upper()
)

ip_country_map = {
    'UK': 'UNITED KINGDOM',
    'U.K.': 'UNITED KINGDOM',
    'GB': 'UNITED KINGDOM',
    'KE': 'KENYA',
    'US': 'UNITED STATES',
    'USA': 'UNITED STATES'
}

cleaned_customer_transaction_data['ip_country'] = (
    cleaned_customer_transaction_data['ip_country']
    .replace(ip_country_map)
)

# Fill missing ip_country using home_country
cleaned_customer_transaction_data['ip_country'] = (
    cleaned_customer_transaction_data['ip_country']
    .fillna(cleaned_customer_transaction_data['home_country'])
)
```

```python
# --- Normalize kyc_tier ---

cleaned_customer_transaction_data['kyc_tier'] = (
    cleaned_customer_transaction_data['kyc_tier']
    .astype(str)
    .str.strip()
    .str.lower()
)

kyc_map = {
    'tier1': 'Tier 1',
    'tier 1': 'Tier 1',
    't1': 'Tier 1',
    'basic': 'Tier 1',

    'tier2': 'Tier 2',
    'tier 2': 'Tier 2',
    't2': 'Tier 2',
    'standard': 'Tier 2',

    'tier3': 'Tier 3',
    'tier 3': 'Tier 3',
    't3': 'Tier 3',
    'enhanced': 'Tier 3'
}

cleaned_customer_transaction_data['kyc_tier'] = (
    cleaned_customer_transaction_data['kyc_tier']
    .replace(kyc_map)
)

# Fill missing with mode
cleaned_customer_transaction_data['kyc_tier'] = (
    cleaned_customer_transaction_data['kyc_tier']
    .fillna(cleaned_customer_transaction_data['kyc_tier'].mode()[0])
)

cleaned_customer_transaction_data['kyc_tier'] = cleaned_customer_transaction_data['kyc_tier'].astype('category')
```

What the feature does
The feature compares:

home_country → where the customer is registered

ip_country → where the transaction originated from (based on IP address)

If they are different, we mark the transaction as 1 (mismatch).
If they are the same, we mark it as 0 (no mismatch).

Why I used it-I created the country_mismatch feature because location inconsistency is a well‑known fraud indicator.
It helps the fraud detection model identify:

suspicious login locations

abnormal transaction origins

potential account compromise

high‑risk cross‑border behaviour
