Distribution of Fraud and Transaction Channels

Understanding the distribution of fraud and transaction channels is essential for exploratory data analysis (EDA) and machine learning model development.
The fraud distribution chart helps identify class imbalance, which affects model performance and evaluation metrics.
The transaction channel distribution chart highlights dominant channels and potential patterns linked to fraudulent behavior.
These observations guide feature analysis and model development.

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
```

```python
# Fraud distribution
sns.countplot(data=cleaned_customer_transaction_data, x='is_fraud', ax=axes[0])
axes[0].set_title('Fraud Transaction Distribution')
axes[0].set_xlabel('Fraud Status')
axes[0].set_ylabel('Number of Transactions')
axes[0].set_xticks([0, 1])
axes[0].set_xticklabels(['Not Fraud', 'Fraud'])

# Channel distribution
sns.countplot(data=cleaned_customer_transaction_data, x='channel', ax=axes[1])
axes[1].set_title('Transaction Channel Distribution')
axes[1].set_xlabel('Channel')
axes[1].set_ylabel('Number of Transactions')

plt.tight_layout()
plt.show()
```
